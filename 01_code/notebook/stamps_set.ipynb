```python
import os 
import glob
import re
```

```python
path = "../../El-picpic-stamp-yolo/labels"
output_path = "../../El-picpic-stamp-yolo/labels_corrected"
label_mapping = {
    0: None,
    1: 3,
}

total_files = 0
total_deleted = 0
total_kept = 0
```

```python
for filename in os.listdir(path):
    total_files += 1

    input_path = os.path.join(path, filename)
    output_dir =  os.path.join(output_path, filename)

    with open(input_path, 'r') as f:
        lines = f.readlines()
    
    new_lines=[]

    for line in lines:
        parts = line.strip().split()
        if not parts:
            continue
        
        original_label = int(parts[0])
        new_label = label_mapping.get(original_label)
        
        if new_label is None:
            total_deleted =+ 1
            continue
        
        parts[0] = str(new_label)
        new_lines.append(" ".join(parts))
        total_kept += 1

    with open(output_dir, "w") as f:
        f.write('\n'.join(new_lines))
        if new_lines:
            f.write('\n')

print(total_files)
print(total_deleted)
print(total_kept)  


```
